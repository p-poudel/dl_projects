Converting RGB images to Grayscale using openCV

```python
import cv2
```

```python
img = cv2.imread('test_image.png')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscaled_image = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
grayscaled_image
```

```python
grayscaled_image.shape
```

```python
plt.imshow(grayscaled_image)
plt.show()
```

```python
plt.imshow(img)
plt.show()
```

```python
cv2.imshow('Grayscale Image', grayscaled_image)
```

```python
cv2.imwrite('Grayscale_Image.jpg', grayscaled_image)
```
