Importing the Dependencies

```python
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

Data Collection and Processing

```python
# loading the data from sklearn
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
breast_cancer_dataset
```

```python
# loading the data to a Pandas Dataframe
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns=breast_cancer_dataset.feature_names)
```

```python
# adding the target column
data_frame['label'] = breast_cancer_dataset.target
```

Splitting Features and Target

```python
X = data_frame.drop('label', axis=1)
Y = data_frame['label']
```

```python
X
```

```python
Y
```

Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,stratify=Y,test_size=0.2,random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Standardize the Data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_std = scaler.fit_transform(X_train) # at first it fits the scaler and transforms the data
X_test_std = scaler.transform(X_test)
```

```python
X_train_std
```

Building the Neural Network

```python
# imporintg tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
# setting up the layers of Neural Network

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)), # Flatten converts the input data into 1-D array
    keras.layers.Dense(20, activation='relu'), # Dense means each neuron in this layer is connected to every neuron in previous layer
    keras.layers.Dense(2, activation='sigmoid') # no of neurons = len(data_frame.label.value_counts())
])
```

```python
# compiling the neural network

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy', # good for interger categorized, for one not, use categorical_crossentropy
              metrics=['accuracy'])
```

```python
# training the neural network

history = model.fit(X_train_std, Y_train, validation_split=0.1, epochs=10)
```

Visualizing the Accuracy and Loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')

plt.legend(['training data', 'validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')

plt.legend(['training data', 'validation data'], loc = 'upper right')
```

Accuracy of the model on test data

```python
loss, accuracy = model.evaluate(X_test_std, Y_test)
```
